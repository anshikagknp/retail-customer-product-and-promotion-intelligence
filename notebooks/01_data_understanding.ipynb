```python
hh_demo = pd.read_csv("../data/raw/hh_demographic.csv")

print("Shape:", hh_demo.shape)
print("\nColumns:", hh_demo.columns.to_list())

display(hh_demo.head())

hh_demo.info()
```

```python
print("Unique households:", hh_demo["household_key"].nunique())
print("Duplicate household IDs:", hh_demo["household_key"].duplicated().sum())
print("Duplicate complete rows:", hh_demo.duplicated().sum())

print("\nMissing values:")
print(hh_demo.isnull().sum())
```

```python
for col in hh_demo.columns:
    if col != "household_key":
        print(f"\n{col}")
        print(hh_demo[col].value_counts())
```

```python
print("Unique transaction households:", df["household_key"].nunique())
print("Unique demographic households:", hh_demo["household_key"].nunique())

print("Demographic households missing from transactions:",
      (~hh_demo["household_key"].isin(df["household_key"])).sum())

print("Transaction households missing from demographics:",
      (~df["household_key"].isin(hh_demo["household_key"])).sum())
```

```python
unmatched = df.loc[
    ~df["household_key"].isin(hh_demo["household_key"]),
    "household_key"
].nunique()

print("Unique transaction households without demographic data:", unmatched)
```

```python
household_check = (
    df[["household_key"]]
    .drop_duplicates()
    .merge(
        hh_demo[["household_key"]],
        on="household_key",
        how="left",
        indicator=True
    )
)

household_check["_merge"].value_counts()
```

```python
df_enriched = df.merge(
    hh_demo,
    on="household_key",
    how="left",
    indicator=True
)

print("Merged shape:", df_enriched.shape)
print(df_enriched["_merge"].value_counts())
print("Missing homeowner values:", df_enriched["HOMEOWNER_DESC"].isna().sum())
```

```python
print("Original rows:", len(df))
print("Merged rows:", len(df_enriched))
print("Duplicate transaction rows:", df_enriched.duplicated(
    subset=["household_key", "BASKET_ID", "PRODUCT_ID", "DAY"]
).sum())
```
