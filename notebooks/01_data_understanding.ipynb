### 4. Promotional Datasets Exploration

Inspecting campaign metadata, household targeting, coupon catalog, and redemption logs.

```python
# 4.1 Campaign Descriptions
camp_desc = pd.read_csv('../data/raw/campaign_desc.csv')
print('Campaign Desc Shape:', camp_desc.shape)
display(camp_desc.head())
print('Campaign Types distribution:')
print(camp_desc['DESCRIPTION'].value_counts())
camp_desc['DURATION_DAYS'] = camp_desc['END_DAY'] - camp_desc['START_DAY']
print('Campaign duration statistics (days):')
display(camp_desc[['START_DAY', 'END_DAY', 'DURATION_DAYS']].describe())
```

```python
# 4.2 Campaign Household Targeting Table
camp_table = pd.read_csv('../data/raw/campaign_table.csv')
print('Campaign Table Shape:', camp_table.shape)
print('Unique targeted households:', camp_table['household_key'].nunique())
print('Unique campaigns:', camp_table['CAMPAIGN'].nunique())
print('Campaigns per household summary:')
display(camp_table.groupby('household_key')['CAMPAIGN'].count().describe())
display(camp_table.head())
```

```python
# 4.3 Coupon Catalog
coupon = pd.read_csv('../data/raw/coupon.csv')
print('Coupon Table Shape:', coupon.shape)
print('Unique coupon UPCs:', coupon['COUPON_UPC'].nunique())
print('Unique products linked to coupons:', coupon['PRODUCT_ID'].nunique())
print('Duplicate rows in coupon.csv:', coupon.duplicated().sum())
display(coupon.head())
```

```python
# 4.4 Coupon Redemptions
coupon_redempt = pd.read_csv('../data/raw/coupon_redempt.csv')
print('Coupon Redemptions Shape:', coupon_redempt.shape)
print('Unique households redeeming coupons:', coupon_redempt['household_key'].nunique())
print('Unique coupons redeemed:', coupon_redempt['COUPON_UPC'].nunique())
print('Unique campaigns with redemptions:', coupon_redempt['CAMPAIGN'].nunique())
print('Redemption day range:', coupon_redempt['DAY'].min(), 'to', coupon_redempt['DAY'].max())
display(coupon_redempt.head())
```

### 5. Cross-Table Referential Integrity & Quality Assessment

Validating relationships across transactions, products, demographics, campaigns, coupons, and redemptions.

```python
# 5.1 Foreign key and relationship validations
print('1. Product referential integrity:')
unmatched_products = (~df['PRODUCT_ID'].isin(product['PRODUCT_ID'])).sum()
print(f'   Transactions with invalid PRODUCT_ID: {unmatched_products}')

print('2. Campaign referential integrity:')
unmatched_camp = (~camp_table['CAMPAIGN'].isin(camp_desc['CAMPAIGN'])).sum()
print(f'   Campaign table rows with invalid CAMPAIGN: {unmatched_camp}')

print('3. Coupon-Product referential integrity:')
coupon_prod_unmatched = (~coupon['PRODUCT_ID'].isin(product['PRODUCT_ID'])).sum()
print(f'   Coupons referencing invalid PRODUCT_ID: {coupon_prod_unmatched}')

print('4. Redemption targeting validity:')
target_pairs = set(zip(camp_table['household_key'], camp_table['CAMPAIGN']))
redempt_pairs = list(zip(coupon_redempt['household_key'], coupon_redempt['CAMPAIGN']))
valid_redempts = sum(pair in target_pairs for pair in redempt_pairs)
print(f'   Redemptions matching campaign household targeting: {valid_redempts} / {len(coupon_redempt)} ({valid_redempts/len(coupon_redempt)*100:.1f}%)')

print('5. Redemption coupon validity:')
coupon_pairs = set(zip(coupon['COUPON_UPC'], coupon['CAMPAIGN']))
redempt_coupon_pairs = list(zip(coupon_redempt['COUPON_UPC'], coupon_redempt['CAMPAIGN']))
valid_coupon_redempts = sum(pair in coupon_pairs for pair in redempt_coupon_pairs)
print(f'   Redemptions matching coupon catalog: {valid_coupon_redempts} / {len(coupon_redempt)} ({valid_coupon_redempts/len(coupon_redempt)*100:.1f}%)')
```
